# Optimize driving efficiency — **Ridge** regression

One regressor setup (with **StandardScaler** fit on each training split); **three datasets** (DS2, DS4, DS5).

**Run the first code cell** so imports resolve. Kernel working directory should be the **repository root**, or any folder from which walking up finds `data/processed/`.

```python
from pathlib import Path
import sys

_cwd = Path.cwd().resolve()
PROJECT_ROOT = None
for p in [_cwd, *_cwd.parents]:
    if (p / "data" / "processed").is_dir():
        PROJECT_ROOT = p
        break
if PROJECT_ROOT is None:
    raise FileNotFoundError("Could not find data/processed — open/run from the repo root.")

_PKG = PROJECT_ROOT / "model_implementation" / "optimize_driving_efficiency"
sys.path.insert(0, str(_PKG))

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

from common_metrics import print_metrics
from ds2_ev_specs_data import TARGET_DESC as DS2_TARGET, load_splits as load_ds2
from ds4_ev_charging_data import TARGET_DESC as DS4_TARGET, load_splits as load_ds4
from ds5_ev_energy_data import TARGET_DESC as DS5_TARGET, load_splits as load_ds5

print("Project root:", PROJECT_ROOT)
```

## Dataset 2 — EV specs → **Efficiency (Wh/km)**

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds2()

scaler_ds2 = StandardScaler()
X_tr_s = scaler_ds2.fit_transform(X_tr)
X_va_s = scaler_ds2.transform(X_va)
X_te_s = scaler_ds2.transform(X_te)

model_ds2 = Ridge(alpha=1.0)
model_ds2.fit(X_tr_s, y_tr)

print_metrics(
    "Ridge — DS2 EV specs",
    y_va,
    model_ds2.predict(X_va_s),
    split_name="Validation",
    target_desc=DS2_TARGET,
)
print_metrics(
    "Ridge — DS2 EV specs",
    y_te,
    model_ds2.predict(X_te_s),
    split_name="Test",
    target_desc=DS2_TARGET,
)
```

## Dataset 4 — EV charging → **Charging Cost (USD)**

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds4()

scaler_ds4 = StandardScaler()
X_tr_s = scaler_ds4.fit_transform(X_tr)
X_va_s = scaler_ds4.transform(X_va)
X_te_s = scaler_ds4.transform(X_te)

model_ds4 = Ridge(alpha=1.0)
model_ds4.fit(X_tr_s, y_tr)

print_metrics(
    "Ridge — DS4 EV charging",
    y_va,
    model_ds4.predict(X_va_s),
    split_name="Validation",
    target_desc=DS4_TARGET,
)
print_metrics(
    "Ridge — DS4 EV charging",
    y_te,
    model_ds4.predict(X_te_s),
    split_name="Test",
    target_desc=DS4_TARGET,
)
```

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds5()

scaler_ds5 = StandardScaler()
X_tr_s = scaler_ds5.fit_transform(X_tr)
X_va_s = scaler_ds5.transform(X_va)
X_te_s = scaler_ds5.transform(X_te)

model_ds5 = Ridge(alpha=1.0)
model_ds5.fit(X_tr_s, y_tr)

print_metrics(
    "Ridge — DS5 EV energy consumption",
    y_va,
    model_ds5.predict(X_va_s),
    split_name="Validation",
    target_desc=DS5_TARGET,
)
print_metrics(
    "Ridge — DS5 EV energy consumption",
    y_te,
    model_ds5.predict(X_te_s),
    split_name="Test",
    target_desc=DS5_TARGET,
)
```
