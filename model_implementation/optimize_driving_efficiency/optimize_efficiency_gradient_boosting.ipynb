# Optimize driving efficiency — **Gradient Boosting**

One regressor setup; **three datasets** (DS2, DS4, DS5) with the targets and feature sets defined for this use case.

**Run the first code cell** so imports resolve. Kernel working directory should be the **repository root**, or any folder from which walking up finds `data/processed/`.

```python
from pathlib import Path
import sys

_cwd = Path.cwd().resolve()
PROJECT_ROOT = None
for p in [_cwd, *_cwd.parents]:
    if (p / "data" / "processed").is_dir():
        PROJECT_ROOT = p
        break
if PROJECT_ROOT is None:
    raise FileNotFoundError("Could not find data/processed — open/run from the repo root.")

_PKG = PROJECT_ROOT / "model_implementation" / "optimize_driving_efficiency"
sys.path.insert(0, str(_PKG))

from sklearn.ensemble import GradientBoostingRegressor

from common_metrics import print_metrics
from ds2_ev_specs_data import TARGET_DESC as DS2_TARGET, load_splits as load_ds2
from ds4_ev_charging_data import TARGET_DESC as DS4_TARGET, load_splits as load_ds4
from ds5_ev_energy_data import TARGET_DESC as DS5_TARGET, load_splits as load_ds5

print("Project root:", PROJECT_ROOT)
```

## Dataset 2 — EV specs → **Efficiency (Wh/km)**

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds2()

model_ds2 = GradientBoostingRegressor(
    n_estimators=200, learning_rate=0.1, max_depth=3, random_state=42
)
model_ds2.fit(X_tr, y_tr)

print_metrics(
    "Gradient Boosting — DS2 EV specs",
    y_va,
    model_ds2.predict(X_va),
    split_name="Validation",
    target_desc=DS2_TARGET,
)
print_metrics(
    "Gradient Boosting — DS2 EV specs",
    y_te,
    model_ds2.predict(X_te),
    split_name="Test",
    target_desc=DS2_TARGET,
)
```

## Dataset 4 — EV charging → **Charging Cost (USD)**

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds4()

model_ds4 = GradientBoostingRegressor(
    n_estimators=200, learning_rate=0.1, max_depth=3, random_state=42
)
model_ds4.fit(X_tr, y_tr)

print_metrics(
    "Gradient Boosting — DS4 EV charging",
    y_va,
    model_ds4.predict(X_va),
    split_name="Validation",
    target_desc=DS4_TARGET,
)
print_metrics(
    "Gradient Boosting — DS4 EV charging",
    y_te,
    model_ds4.predict(X_te),
    split_name="Test",
    target_desc=DS4_TARGET,
)
```

## Dataset 5 — EV energy consumption → **Energy_Consumption_kWh**

```python
X_tr, X_va, X_te, y_tr, y_va, y_te = load_ds5()

model_ds5 = GradientBoostingRegressor(
    n_estimators=200, learning_rate=0.1, max_depth=3, random_state=42
)
model_ds5.fit(X_tr, y_tr)

print_metrics(
    "Gradient Boosting — DS5 EV energy consumption",
    y_va,
    model_ds5.predict(X_va),
    split_name="Validation",
    target_desc=DS5_TARGET,
)
print_metrics(
    "Gradient Boosting — DS5 EV energy consumption",
    y_te,
    model_ds5.predict(X_te),
    split_name="Test",
    target_desc=DS5_TARGET,
)
```
